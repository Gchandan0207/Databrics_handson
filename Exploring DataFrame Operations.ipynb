```python
df = spark.read.csv(
    path="/Volumes/quickstart_catalog/quickstart_schema/sandbox/dataset/employee.csv",
    header=True,
    inferSchema=True,
    sep="|",
    quote="'"
)
df.display()
```

```python
df.select("id","name").display()
```

```python
from pyspark.sql.functions import col
#recommonded way of calling an column using column object
#advantage is that you can use dot notation alias 
df.select(col("id"), col("name").alias("full_name")).display()
```

#Filter Operations

```python
df.filter(col("company")=="TCS").display()
```

```python
df.where(col("company")=="TCS").display()
```

```python
df.select(col("company")=="TCS").display()
#it gives boolean value
```

```python
df.groupBy("gen").count().display()
```

```python
from pyspark.sql.functions import max, min, avg
df.groupBy("gen").agg(max("exp")).display()
```

```python
from pyspark.sql.functions import max, min, avg
df.groupBy("gen").agg(max("exp").alias("max_exp"),min("exp").alias("min_exp")).display()
```
