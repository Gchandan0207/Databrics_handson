#Prerequisite

```python
df=spark.read.csv(path="/Volumes/quickstart_catalog/quickstart_schema/sandbox/dataset/user_dataset/users_001.csv",header=True,inferSchema=True)
display(df)
```

```python
#write it as parquet format
df.write.format("parquet").save("/Volumes/quickstart_catalog/quickstart_schema/sandbox/output_parquet")
```

```python
#read from parquet format
df_parquet = spark.read.parquet("/Volumes/quickstart_catalog/quickstart_schema/sandbox/output_parquet")
display(df_parquet)
```

#Create View
##Note-To run sql queries on DF we need to create views


```python
df_parquet.createOrReplaceTempView("user_vw")
```

```python
%sql
UPDATE user_vw
set country = 'Bharat'
where country = 'India'
```

```python
from pyspark.sql.functions import col
df.filter(col("country") == "India").write.save(
    format="parquet",
    mode="OVERWRITE",
    path="/Volumes/quickstart_catalog/quickstart_schema/sandbox/output_parquet",
)
```
