#Prerequisite

```python
df=spark.read.csv(path="/Volumes/quickstart_catalog/quickstart_schema/sandbox/dataset/user_dataset/users_001.csv",header=True,inferSchema=True)
display(df)
```

#Transaction 01 - Write the data as Delta Format --(ACID comes at the cost of storage but now a days storage is cheap)

```python
df.write.save(
    format="delta",#by default it is delta
    mode="overwrite",
    path="/Volumes/quickstart_catalog/quickstart_schema/sandbox/output_delta",
)
```

Transaction 01 - Overwrite Delta Folder

```python
from pyspark.sql.functions import col
df.filter(col("country") == "India").write.format("delta").mode("overwrite").save(
    path="/Volumes/quickstart_catalog/quickstart_schema/sandbox/output_delta"
)
```

#Read Delta - Latest Transaction

```python
spark.read.format("delta").load(
    path="/Volumes/quickstart_catalog/quickstart_schema/sandbox/output_delta"
).display()
```

##Observaton - After this it will create new parquet file while keeping the old one alongg with new json file for new transaction in __delta_log folder

#Maintains Versions
## we can read older versions of data

```python
spark.read.option("versionAsof",0).load(
    path="/Volumes/quickstart_catalog/quickstart_schema/sandbox/output_delta"
).display()
#it will read records with us and indida  
```

```python
spark.read.option("versionAsof",1).load(
    path="/Volumes/quickstart_catalog/quickstart_schema/sandbox/output_delta"
).display()
#it will read records with onlyindida  
```
